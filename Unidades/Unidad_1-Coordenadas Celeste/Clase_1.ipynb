# ¿Qué es un archivo FITS?

**FITS** (*Flexible Image Transport System*) es el formato estándar utilizado por la comunidad astronómica para almacenar datos científicos.

A diferencia de formatos como JPG o PNG, un archivo FITS almacena no solo la imagen, sino también toda la información necesaria para interpretar una observación:

- Instrumento
- Telescopio
- Fecha de observación
- Tiempo de exposición
- Filtro
- Coordenadas del objeto
- Datos científicos

> **Concepto:** Un archivo FITS es un contenedor de datos científicos, no simplemente una imagen.


# Estructura de un FITS

Todo archivo FITS está compuesto por una o más **HDU (Header Data Unit)**.

```
Archivo FITS
│
├── Primary HDU
│   ├── Header
│   └── Data
│
├── Extension HDU
│   ├── Header
│   └── Data
```

El **Header** contiene metadatos y la **Data** contiene la imagen, espectro o tabla.

# ¿Qué es SIMBAD?

SIMBAD significa **Set of Identifications, Measurements, and Bibliography for Astronomical Data**.

Su objetivo principal es proporcionar información sobre objetos astronómicos que han sido estudiados en publicaciones científicas.

Entre la información disponible podemos encontrar:

- Identificadores y nombres alternativos.
- Coordenadas.
- Magnitudes.
- Tipo de objeto.
- Tipo espectral o morfológico.
- Movimiento propio.
- Paralaje.
- Velocidad o redshift.
- Bibliografía.

> **Importante:** SIMBAD no debe confundirse con un catálogo homogéneo como Gaia. SIMBAD es una **meta-compilación** de información procedente de la literatura y de diferentes catálogos


https://simbad.cds.unistra.fr/simbad/

## ¿Qué es Astroquery?

En lugar de copiar manualmente información desde una página web, podemos automatizar la consulta desde Python.

La idea general es:

```text
Python
   ↓
Astroquery
   ↓
SIMBAD
   ↓
Resultado
   ↓
Astropy Table
```

Esto permite construir análisis reproducibles y consultar grandes cantidades de información sin tener que realizar cada búsqueda manualmente.

## Buenas prácticas al consultar SIMBAD
### 1. No hagas miles de consultas individuales
Evita:

```python

    Simbad.query_object(obj)
```
si puedes realizar una consulta conjunta.

Utiliza `query_objects()` para una lista de nombres.

### 2. No supongas que SIMBAD es un catálogo homogéneo

### 3. Comprueba las unidades

### 4. Comprueba los identificadores
